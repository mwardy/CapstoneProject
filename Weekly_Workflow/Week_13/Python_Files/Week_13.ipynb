```python
import numpy as np
import warnings

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, ConstantKernel, WhiteKernel
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error

print("=" * 70)
print("STEP 3 – GP AND NN MODEL DIAGNOSTICS")
print("=" * 70)

gp_models = {}
nn_models = {}

cv = KFold(n_splits=5, shuffle=True, random_state=42)

for i in range(1, 9):

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    # Standardise outputs
    y_mean = np.mean(y)
    y_std = np.std(y)

    if y_std == 0:
        y_std = 1.0

    y_scaled = (y - y_mean) / y_std

    # Gaussian Process kernel
    kernel = (
        ConstantKernel(1.0, (1e-3, 1e3))
        * Matern(
            length_scale=np.ones(X.shape[1]),
            length_scale_bounds=(1e-3, 1e3),
            nu=2.5
        )
        + WhiteKernel(
            noise_level=1e-5,
            noise_level_bounds=(1e-8, 1e-1)
        )
    )

    gp = GaussianProcessRegressor(
        kernel=kernel,
        normalize_y=False,
        n_restarts_optimizer=5,
        random_state=42
    )

    # NN model
    nn = make_pipeline(
        StandardScaler(),
        MLPRegressor(
            hidden_layer_sizes=(64, 64),
            activation="relu",
            alpha=0.01,
            max_iter=3000,
            random_state=42
        )
    )

    # Cross-validation for both models
    gp_errors = []
    nn_errors = []

    for train_idx, test_idx in cv.split(X):

        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y_scaled[train_idx], y_scaled[test_idx]

        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            gp.fit(X_train, y_train)

        nn.fit(X_train, y_train)

        gp_pred = gp.predict(X_test)
        nn_pred = nn.predict(X_test)

        gp_errors.extend(abs(y_test - gp_pred))
        nn_errors.extend(abs(y_test - nn_pred))

    gp_cv_mae = np.mean(gp_errors)
    nn_cv_mae = np.mean(nn_errors)

    # Refit models using all available observations
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")
        gp.fit(X, y_scaled)

    nn.fit(X, y_scaled)

    gp_models[i] = (gp, y_mean, y_std)
    nn_models[i] = (nn, y_mean, y_std)

    print(f"\nFUNCTION {i}")
    print(f"Observations: {len(y)}")
    print(f"GP fitted kernel: {gp.kernel_}")
    print(f"GP CV MAE (standardised): {gp_cv_mae:.4f}")
    print(f"NN CV MAE (standardised): {nn_cv_mae:.4f}")

    if gp_cv_mae < nn_cv_mae:
        print("Better validation performance: GP")
    elif nn_cv_mae < gp_cv_mae:
        print("Better validation performance: NN")
    else:
        print("Better validation performance: Equal")

    if caught:
        print("GP fitting warnings:")
        for warning in caught:
            print(f"  - {warning.message}")
    else:
        print("GP fitting warnings: None")
```

```python
import numpy as np

from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.neighbors import NearestNeighbors

print("=" * 70)
print("STEP 5 – PCA, CLUSTERING AND NEIGHBOURHOOD EVIDENCE")
print("=" * 70)

for i in range(1, 9):

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    # PCA on input observations
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    pca = PCA()
    scores = pca.fit_transform(X_scaled)

    best_idx = np.argmax(y)
    top_count = max(3, int(np.ceil(len(y) * 0.20)))
    top_idx = np.argsort(y)[-top_count:]

    # Correlation between PCA scores and outputs
    correlations = []

    for j in range(X.shape[1]):
        corr = np.corrcoef(scores[:, j], y)[0, 1]
        correlations.append(corr)

    strongest_pc = np.argmax(np.abs(correlations))

    # K-Means clustering: test 2 to 4 clusters
    best_silhouette = -1
    best_labels = None
    best_k = None

    for k in range(2, min(5, len(X) - 1)):

        km = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        labels = km.fit_predict(X_scaled)
        score = silhouette_score(X_scaled, labels)

        if score > best_silhouette:
            best_silhouette = score
            best_labels = labels
            best_k = k

    # Cluster containing the current best
    best_cluster = best_labels[best_idx]
    cluster_idx = np.where(best_labels == best_cluster)[0]

    # Nearest observations to the current best
    neighbours = NearestNeighbors(
        n_neighbors=min(6, len(X))
    )
    neighbours.fit(X)

    distances, indices = neighbours.kneighbors(
        X[best_idx].reshape(1, -1)
    )

    # Exclude the best observation itself
    neighbour_indices = indices[0][1:]
    neighbour_distances = distances[0][1:]
    neighbour_outputs = y[neighbour_indices]

    print(f"\nFUNCTION {i}")
    print(f"Observations: {len(y)}")
    print(f"Best output: {y[best_idx]:.10f}")

    print(
        f"PC1 and PC2 variance explained: "
        f"{np.sum(pca.explained_variance_ratio_[:2]):.2%}"
    )

    print(
        f"Strongest PC-output correlation: "
        f"PC{strongest_pc + 1} = "
        f"{correlations[strongest_pc]:.4f}"
    )

    print(
        f"Best observation PC1/PC2 scores: "
        f"{np.round(scores[best_idx, :2], 4)}"
    )

    print(
        f"Top 20% mean PC1/PC2 scores: "
        f"{np.round(np.mean(scores[top_idx, :2], axis=0), 4)}"
    )

    print(f"Best K-Means cluster count: {best_k}")
    print(f"Silhouette score: {best_silhouette:.4f}")
    print(f"Observations in best point's cluster: {len(cluster_idx)}")
    print(f"Mean output in that cluster: {np.mean(y[cluster_idx]):.10f}")

    print(
        f"Mean distance of 5 nearest neighbours: "
        f"{np.mean(neighbour_distances):.4f}"
    )

    print(
        f"Mean output of 5 nearest neighbours: "
        f"{np.mean(neighbour_outputs):.10f}"
    )

    print(
        f"Best output among 5 nearest neighbours: "
        f"{np.max(neighbour_outputs):.10f}"
    )
```

```python
import numpy as np
from sklearn.neighbors import NearestNeighbors

# Cross-validation errors from Step 3
gp_cv_errors = {
    1: 0.5204, 2: 0.6073, 3: 0.4965, 4: 0.1486,
    5: 0.1507, 6: 0.3412, 7: 0.2809, 8: 0.0766
}

nn_cv_errors = {
    1: 0.8246, 2: 0.6159, 3: 0.5004, 4: 0.1977,
    5: 0.1422, 6: 0.3575, 7: 0.3462, 8: 0.2796
}

print("=" * 70)
print("STEP 6 – EXPLORATION–EXPLOITATION ASSESSMENT")
print("=" * 70)

for i in range(1, 9):

    df = data[i]
    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    best_idx = np.argmax(y)
    best_output = y[best_idx]

    # Recent optimisation progress
    observations_since_best = len(y) - best_idx - 1
    recent_outputs = y[-5:]
    recent_best = np.max(recent_outputs)

    # Neighbourhood around current best
    neighbours = NearestNeighbors(
        n_neighbors=min(6, len(X))
    )
    neighbours.fit(X)

    distances, indices = neighbours.kneighbors(
        X[best_idx].reshape(1, -1)
    )

    neighbour_idx = indices[0][1:]
    neighbour_outputs = y[neighbour_idx]

    # Compare output gaps relative to the observed output range
    output_range = np.max(y) - np.min(y)

    if output_range == 0:
        output_range = 1.0

    recent_gap = (best_output - recent_best) / output_range

    neighbour_gap = (
        best_output - np.mean(neighbour_outputs)
    ) / output_range

    # Better cross-validation model
    better_model = (
        "GP" if gp_cv_errors[i] < nn_cv_errors[i]
        else "NN"
    )

    print(f"\nFUNCTION {i}")
    print(f"Observations since best: {observations_since_best}")
    print(f"Recent gap (normalised): {recent_gap:.4f}")
    print(f"Neighbour gap (normalised): {neighbour_gap:.4f}")
    print(f"Better CV model: {better_model}")

    # Preliminary strategy based on previous analysis
    strategies = {
        1: "Limited exploration / model-independent search",
        2: "NN-supported local refinement",
        3: "Conservative local refinement",
        4: "Cautious local refinement around best observations",
        5: "Targeted boundary exploitation",
        6: "GP–NN comparison with PCA-guided refinement",
        7: "GP-led local refinement",
        8: "GP-led local refinement with PCA support"
    }

    print(f"Proposed final strategy: {strategies[i]}")
```

```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import cdist

rng = np.random.default_rng(42)

print("=" * 70)
print("STEP 7 – FUNCTION-SPECIFIC CANDIDATE GENERATION")
print("=" * 70)

# Search strategies from Step 6
strategies = {
    1: "Exploration",
    2: "NN-supported local",
    3: "Conservative local",
    4: "Cautious local",
    5: "Boundary refinement",
    6: "PCA and model comparison",
    7: "GP-led local",
    8: "GP-led local with PCA"
}

# Store candidates for the following steps
candidate_sets = {}

for i in range(1, 9):

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    dimensions = X.shape[1]

    gp, gp_mean, gp_std = gp_models[i]
    nn, nn_mean, nn_std = nn_models[i]

    best_idx = np.argmax(y)
    best_point = X[best_idx]

    candidates = []
    sources = []

    # 1. Local refinement around the five best observations
    top_indices = np.argsort(y)[-5:]

    if i in [3, 4, 7, 8]:
        radii = [0.01, 0.025, 0.05]
    else:
        radii = [0.02, 0.05, 0.10]

    for idx in top_indices:
        centre = X[idx]

        for radius in radii:

            points = centre + rng.normal(
                0, radius, size=(1000, dimensions)
            )

            points = np.clip(points, 0, 1)

            candidates.extend(points)
            sources.extend(["Local"] * len(points))

    # 2. PCA-guided refinement
    if i in [2, 4, 5, 6, 7, 8]:

        scaler = StandardScaler()
        X_scaled = scaler.fit_transform(X)

        pca = PCA()
        scores = pca.fit_transform(X_scaled)

        correlations = [
            np.corrcoef(scores[:, j], y)[0, 1]
            for j in range(dimensions)
        ]

        strongest_pc = np.argmax(np.abs(correlations))

        direction = (
            pca.components_[strongest_pc]
            * scaler.scale_
        )

        direction = direction / (
            np.linalg.norm(direction) + 1e-12
        )

        # Move in direction associated with higher outputs
        if correlations[strongest_pc] < 0:
            direction = -direction

        for step_size in [0.01, 0.025, 0.05, 0.10]:

            for _ in range(250):

                point = (
                    best_point
                    + step_size * direction
                    + rng.normal(0, step_size / 3, dimensions)
                )

                candidates.append(np.clip(point, 0, 1))
                sources.append("PCA")

    # 3. Boundary-focused refinement for F5
    if i == 5:

        for radius in [0.005, 0.015, 0.03]:

            points = 1 - np.abs(
                rng.normal(0, radius, size=(2000, dimensions))
            )

            points = np.clip(points, 0, 1)

            candidates.extend(points)
            sources.extend(["Boundary"] * len(points))

    # 4. Model-independent exploration for F1
    if i == 1:

        points = rng.uniform(
            0, 1, size=(10000, dimensions)
        )

        candidates.extend(points)
        sources.extend(["Exploration"] * len(points))

    candidates = np.asarray(candidates)
    sources = np.asarray(sources)

    # Remove candidates too close to existing observations
    distances = cdist(candidates, X).min(axis=1)

    min_distance = 0.005 if i != 1 else 0.05
    valid = distances >= min_distance

    candidates = candidates[valid]
    sources = sources[valid]
    distances = distances[valid]

    # GP predictions, converted back to original output scale
    gp_scaled_pred, gp_scaled_uncertainty = gp.predict(
        candidates, return_std=True
    )

    gp_pred = gp_scaled_pred * gp_std + gp_mean
    gp_uncertainty = gp_scaled_uncertainty * gp_std

    # NN predictions, converted back to original output scale
    nn_pred = nn.predict(candidates) * nn_std + nn_mean

    # Store everything for Step 8
    candidate_sets[i] = {
        "X": candidates,
        "source": sources,
        "distance": distances,
        "gp_pred": gp_pred,
        "gp_std": gp_uncertainty,
        "nn_pred": nn_pred
    }

    # Find the best candidate according to each model
    gp_best = np.argmax(gp_pred)
    nn_best = np.argmax(nn_pred)

    print(f"\nFUNCTION {i}")
    print(f"Strategy: {strategies[i]}")
    print(f"Candidates generated: {len(candidates)}")
    print(f"Best observed output: {np.max(y):.10f}")

    print("Best GP candidate:")
    print(f"  Source: {sources[gp_best]}")
    print(f"  Point: {np.round(candidates[gp_best], 6)}")
    print(f"  GP prediction: {gp_pred[gp_best]:.10f}")
    print(f"  NN prediction: {nn_pred[gp_best]:.10f}")
    print(f"  Distance: {distances[gp_best]:.4f}")

    print("Best NN candidate:")
    print(f"  Source: {sources[nn_best]}")
    print(f"  Point: {np.round(candidates[nn_best], 6)}")
    print(f"  GP prediction: {gp_pred[nn_best]:.10f}")
    print(f"  NN prediction: {nn_pred[nn_best]:.10f}")
    print(f"  Distance: {distances[nn_best]:.4f}")
```

```python
import numpy as np

print("=" * 70)
print("STEP 8 – CANDIDATE COMPARISON")
print("=" * 70)

# Initial model weights based on Steps 2 and 3
# Each pair is (GP weight, NN weight)
model_weights = {
    1: (0.70, 0.30),
    2: (0.40, 0.60),
    3: (0.50, 0.50),
    4: (0.65, 0.35),
    5: (0.60, 0.40),
    6: (0.50, 0.50),
    7: (0.75, 0.25),
    8: (0.85, 0.15)
}

# Store shortlisted candidates for Step 9
shortlists = {}

for i in range(1, 9):

    results = candidate_sets[i]

    X_candidates = results["X"]
    sources = results["source"]
    distances = results["distance"]
    gp_pred = results["gp_pred"]
    gp_uncertainty = results["gp_std"]
    nn_pred = results["nn_pred"]

    y = data[i].iloc[:, -1].values
    best_observed = np.max(y)

    gp_weight, nn_weight = model_weights[i]

    # Weighted prediction
    weighted_pred = (
        gp_weight * gp_pred
        + nn_weight * nn_pred
    )

    # Model disagreement
    disagreement = np.abs(gp_pred - nn_pred)

    # Predicted improvement over current best
    predicted_improvement = weighted_pred - best_observed

    # Identify candidates
    best_weighted_idx = np.argmax(weighted_pred)
    best_gp_idx = np.argmax(gp_pred)
    best_nn_idx = np.argmax(nn_pred)

    # Find strongest model agreement among candidates
    # with a weighted prediction in the top 10%
    threshold = np.percentile(weighted_pred, 90)
    strong_candidates = np.where(weighted_pred >= threshold)[0]

    best_agreement_idx = strong_candidates[
        np.argmin(disagreement[strong_candidates])
    ]

    selected_indices = {
        "Weighted": best_weighted_idx,
        "GP": best_gp_idx,
        "NN": best_nn_idx,
        "Agreement": best_agreement_idx
    }

    shortlists[i] = {}

    print(f"\nFUNCTION {i}")
    print(f"Current best: {best_observed:.10f}")
    print(f"Model weights: GP={gp_weight:.2f}, NN={nn_weight:.2f}")

    for name, idx in selected_indices.items():

        shortlists[i][name] = {
            "point": X_candidates[idx],
            "source": sources[idx],
            "gp_pred": gp_pred[idx],
            "gp_std": gp_uncertainty[idx],
            "nn_pred": nn_pred[idx],
            "weighted_pred": weighted_pred[idx],
            "disagreement": disagreement[idx],
            "distance": distances[idx],
            "improvement": predicted_improvement[idx]
        }

        print(f"\n{name} candidate:")
        print(f"  Source: {sources[idx]}")
        print(f"  Point: {np.round(X_candidates[idx], 6)}")
        print(f"  GP prediction: {gp_pred[idx]:.10f}")
        print(f"  GP uncertainty: {gp_uncertainty[idx]:.10f}")
        print(f"  NN prediction: {nn_pred[idx]:.10f}")
        print(f"  Weighted prediction: {weighted_pred[idx]:.10f}")
        print(f"  Model disagreement: {disagreement[idx]:.10f}")
        print(f"  Distance: {distances[idx]:.4f}")
        print(f"  Predicted improvement: {predicted_improvement[idx]:.10f}")
```

```python
import numpy as np
from scipy.spatial.distance import cdist

print("=" * 70)
print("STEP 9 – FINAL CANDIDATE SELECTION")
print("=" * 70)

# Preferred candidates based on Step 8 analysis
preferred_candidates = {
    1: "Agreement",
    2: "Weighted",
    3: "Weighted",
    4: "Agreement",
    5: "Weighted",
    6: "Weighted",
    7: "Weighted",
    8: "Agreement"
}

# Store final selections
final_candidates = {}

for i in range(1, 9):

    X = data[i].iloc[:, :-1].values
    y = data[i].iloc[:, -1].values

    best_observed = np.max(y)
    output_range = np.ptp(y)

    if output_range == 0:
        output_range = 1.0

    print(f"\nFUNCTION {i}")
    print(f"Current best: {best_observed:.10f}")

    # Compare the four shortlisted alternatives
    print("\nCandidate comparison:")

    for name, candidate in shortlists[i].items():

        disagreement_ratio = (
            candidate["disagreement"] / output_range
        )

        print(
            f"  {name:10s} | "
            f"Weighted: {candidate['weighted_pred']:.6f} | "
            f"Disagreement: {disagreement_ratio:.4f} | "
            f"Distance: {candidate['distance']:.4f}"
        )

    # Select preferred candidate
    selection = preferred_candidates[i]
    chosen = shortlists[i][selection]

    point = np.asarray(chosen["point"]).copy()

    # Check dimensions and input boundaries
    correct_dimensions = len(point) == X.shape[1]
    within_bounds = np.all(
        (point >= 0) & (point <= 1)
    )

    # Round to submission precision
    point = np.round(point, 6)

    # Check for previous observations
    nearest_distance = cdist(
        point.reshape(1, -1), X
    ).min()

    duplicate = nearest_distance < 1e-6

    final_candidates[i] = point

    print(f"\nSelected: {selection}")
    print(f"Point: {point}")
    print(
        f"Weighted prediction: "
        f"{chosen['weighted_pred']:.10f}"
    )
    print(
        f"GP uncertainty: "
        f"{chosen['gp_std']:.10f}"
    )
    print(
        f"Model disagreement: "
        f"{chosen['disagreement']:.10f}"
    )
    print(f"Nearest observation: {nearest_distance:.6f}")

    print(
        "Validation:",
        "PASS" if (
            correct_dimensions
            and within_bounds
            and not duplicate
        ) else "CHECK REQUIRED"
    )

    # Submission format
    formatted = "-".join(
        f"{value:.6f}" for value in point
    )

    print(f"Submission: {formatted}")

print("\n" + "=" * 70)
print("FINAL SUBMISSION CANDIDATES")
print("=" * 70)

for i, point in final_candidates.items():
    formatted = "-".join(
        f"{value:.6f}" for value in point
    )
    print(f"F{i}: {formatted}")
```

```python
import numpy as np
from scipy.stats import norm
from scipy.spatial.distance import cdist

print("=" * 70)
print("STEP 9B – FINAL PI / EI COMPARISON")
print("=" * 70)

# Store candidates for final review
final_review = {}

for i in range(1, 9):

    results = candidate_sets[i]

    X_candidates = results["X"]
    gp_pred = results["gp_pred"]
    gp_std = results["gp_std"]
    nn_pred = results["nn_pred"]
    distances = results["distance"]
    sources = results["source"]

    y = data[i].iloc[:, -1].values
    best_observed = np.max(y)

    # Avoid division by zero
    sigma = np.maximum(gp_std, 1e-12)

    # Improvement relative to best observed result
    improvement = gp_pred - best_observed

    z = improvement / sigma

    # Probability of Improvement
    pi = norm.cdf(z)

    # Expected Improvement
    ei = (
        improvement * norm.cdf(z)
        + sigma * norm.pdf(z)
    )

    ei = np.maximum(ei, 0)

    # Conservative prediction: GP mean minus uncertainty
    conservative = gp_pred - gp_std

    # Candidate with highest PI
    pi_idx = np.argmax(pi)

    # Candidate with highest EI
    ei_idx = np.argmax(ei)

    # Candidate with highest conservative score
    conservative_idx = np.argmax(conservative)

    # Current selected candidate from Step 9
    current_point = final_candidates[i]

    current_idx = np.argmin(
        cdist(
            current_point.reshape(1, -1),
            X_candidates
        ).ravel()
    )

    indices = {
        "Current": current_idx,
        "PI": pi_idx,
        "EI": ei_idx,
        "Conservative": conservative_idx
    }

    final_review[i] = {}

    print(f"\nFUNCTION {i}")
    print(f"Best observed: {best_observed:.10f}")

    for name, idx in indices.items():

        final_review[i][name] = {
            "point": X_candidates[idx].copy(),
            "source": sources[idx],
            "gp_pred": gp_pred[idx],
            "gp_std": gp_std[idx],
            "nn_pred": nn_pred[idx],
            "pi": pi[idx],
            "ei": ei[idx],
            "conservative": conservative[idx],
            "distance": distances[idx]
        }

        print(f"\n{name} candidate:")
        print(f"  Source: {sources[idx]}")
        print(f"  Point: {np.round(X_candidates[idx], 6)}")
        print(f"  GP prediction: {gp_pred[idx]:.10f}")
        print(f"  NN prediction: {nn_pred[idx]:.10f}")
        print(f"  GP uncertainty: {gp_std[idx]:.10f}")
        print(f"  PI: {pi[idx]:.6f}")
        print(f"  EI: {ei[idx]:.10f}")
        print(f"  Conservative score: {conservative[idx]:.10f}")
        print(f"  Distance: {distances[idx]:.4f}")

print("\n" + "=" * 70)
print("SUMMARY – CURRENT VS EI")
print("=" * 70)

for i in range(1, 9):

    current = final_review[i]["Current"]
    ei_best = final_review[i]["EI"]

    print(f"\nF{i}")
    print(f"Current PI: {current['pi']:.4f}")
    print(f"EI candidate PI: {ei_best['pi']:.4f}")
    print(f"Current EI: {current['ei']:.8f}")
    print(f"Best EI: {ei_best['ei']:.8f}")
    print(
        "EI improvement over current: "
        f"{ei_best['ei'] - current['ei']:.8f}"
    )
```

```python

import numpy as np
import pandas as pd

print("=" * 75)
print("STEP 10 – OVERALL OPTIMISATION PERFORMANCE")
print("=" * 75)

performance_summary = {}

for i in range(1, 9):

    y = data[i].iloc[:, -1].to_numpy(dtype=float)

    # Each function began with 10 initial observations
    initial_count = 10

    initial_y = y[:initial_count]
    initial_best = np.max(initial_y)

    overall_best = np.max(y)
    best_index = np.argmax(y)

    improvement = overall_best - initial_best

    if abs(initial_best) > 1e-10:
        percentage_improvement = (
            improvement / abs(initial_best)
        ) * 100
    else:
        percentage_improvement = np.nan

    # Running best across all observations
    running_best = np.maximum.accumulate(y)

    # Number of improvements after the initial observations
    improvements_after_initial = np.sum(
        running_best[initial_count:] >
        running_best[initial_count - 1:-1] + 1e-12
    )

    observations_since_best = len(y) - best_index - 1

    performance_summary[i] = {
        "initial_best": initial_best,
        "overall_best": overall_best,
        "improvement": improvement,
        "percentage_improvement": percentage_improvement,
        "best_observation": best_index + 1,
        "total_observations": len(y),
        "improvements_after_initial": improvements_after_initial,
        "observations_since_best": observations_since_best
    }

    print(f"\nFUNCTION {i}")
    print(f"Initial best: {initial_best:.10f}")
    print(f"Overall best: {overall_best:.10f}")
    print(f"Absolute improvement: {improvement:.10f}")

    if np.isfinite(percentage_improvement):
        print(
            f"Percentage improvement: "
            f"{percentage_improvement:.2f}%"
        )
    else:
        print("Percentage improvement: N/A (near-zero baseline)")

    print(f"Best found at observation: {best_index + 1}")
    print(f"Total observations: {len(y)}")
    print(
        "Improvements after initial observations: "
        f"{improvements_after_initial}"
    )
    print(
        f"Observations since last best: "
        f"{observations_since_best}"
    )

print("\n" + "=" * 75)
print("OVERALL SUMMARY")
print("=" * 75)

functions_improved = sum(
    v["improvement"] > 1e-12
    for v in performance_summary.values()
)

total_improvements = sum(
    v["improvements_after_initial"]
    for v in performance_summary.values()
)

print(f"Functions improved: {functions_improved}/8")
print(f"New best observations after initial data: {total_improvements}")

print("\nFINAL SUBMISSION POINTS")

# Updated F1 following Step 9b
final_candidates[1] = np.array([0.986278, 0.694632])

for i in range(1, 9):
    point = final_candidates[i]
    submission = "-".join(f"{v:.6f}" for v in point)
    print(f"F{i}: {submission}")
```
