```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF
import warnings
from sklearn.exceptions import ConvergenceWarning

gp_models = {}

for i in range(1, 9):

    X = (
        data[i]
        .drop(columns="output")
        .values
        .astype(float)
    )

    y = (
        data[i]["output"]
        .values
        .astype(float)
    )

    # Same GP structure used previously
    kernel = RBF(
        length_scale=1.0,
        length_scale_bounds=(1e-5, 1e5)
    )

    gp = GaussianProcessRegressor(
        kernel=kernel,
        normalize_y=True,
        n_restarts_optimizer=10,
        random_state=42
    )

    gp.fit(X, y)

    gp_models[i] = gp

    print("=" * 60)
    print(f"FUNCTION {i}")
    print("=" * 60)
    print(f"Fitted kernel: {gp.kernel_}")
    print()
```

```python
rng = np.random.default_rng(42)

diagnostic_candidates = {}
gp_diagnostics = {}

for i in range(1, 9):

    n_dimensions = data[i].shape[1] - 1

    # Generate candidate points across the search space
    candidates = rng.random(
        (100000, n_dimensions)
    )

    diagnostic_candidates[i] = candidates

    # GP predictions and uncertainty
    mean_pred, std_pred = gp_models[i].predict(
        candidates,
        return_std=True
    )

    gp_diagnostics[i] = {
        "mean": mean_pred,
        "std": std_pred
    }

    print("=" * 60)
    print(f"FUNCTION {i}")
    print("=" * 60)

    print(f"Predicted mean minimum: {mean_pred.min():.10f}")
    print(f"Predicted mean maximum: {mean_pred.max():.10f}")
    print(
        f"Predicted mean range:   "
        f"{mean_pred.max() - mean_pred.min():.10f}"
    )

    print()
    print(f"Uncertainty minimum:    {std_pred.min():.10f}")
    print(f"Uncertainty maximum:    {std_pred.max():.10f}")
    print(
        f"Uncertainty range:      "
        f"{std_pred.max() - std_pred.min():.10f}"
    )

    print()
```

```python
ucb_kappas = [0.5, 1.0, 1.5, 2.5]

ucb_tuning_results = {}

for i in range(1, 9):

    candidates = diagnostic_candidates[i]
    mean_pred = gp_diagnostics[i]["mean"]
    std_pred = gp_diagnostics[i]["std"]

    X_existing = (
        data[i]
        .drop(columns="output")
        .values
        .astype(float)
    )

    y_existing = (
        data[i]["output"]
        .values
        .astype(float)
    )

    # Current best observation
    best_index = np.argmax(y_existing)
    best_point = X_existing[best_index]
    best_output = y_existing[best_index]

    ucb_tuning_results[i] = {}

    print("=" * 70)
    print(f"FUNCTION {i}")
    print("=" * 70)
    print(f"Current best output: {best_output:.6f}")
    print()

    for kappa in ucb_kappas:

        # Calculate UCB for every candidate
        ucb_scores = mean_pred + kappa * std_pred

        # Select highest-scoring candidate
        idx = np.argmax(ucb_scores)

        point = candidates[idx]
        predicted = mean_pred[idx]
        uncertainty = std_pred[idx]
        score = ucb_scores[idx]

        # Distance from current best point
        distance_best = np.linalg.norm(
            point - best_point
        )

        # Distance from nearest existing observation
        distances_existing = np.linalg.norm(
            X_existing - point,
            axis=1
        )

        distance_nearest = distances_existing.min()

        ucb_tuning_results[i][kappa] = {
            "point": point,
            "predicted": predicted,
            "uncertainty": uncertainty,
            "score": score,
            "distance_best": distance_best,
            "distance_nearest": distance_nearest
        }

        print(f"κ = {kappa}")
        print(np.round(point, 6))
        print(f"Predicted output:               {predicted:.6f}")
        print(f"Uncertainty:                    {uncertainty:.6f}")
        print(f"UCB score:                      {score:.6f}")
        print(f"Distance from current best:     {distance_best:.4f}")
        print(f"Distance from nearest existing: {distance_nearest:.4f}")
        print()
```

```python
from scipy.stats import norm

xi_proportions = [0.00, 0.01, 0.05, 0.10]

ei_tuning_results = {}

for i in range(1, 9):

    candidates = diagnostic_candidates[i]
    mean_pred = gp_diagnostics[i]["mean"]
    std_pred = gp_diagnostics[i]["std"]

    X_existing = (
        data[i]
        .drop(columns="output")
        .values
        .astype(float)
    )

    y_existing = (
        data[i]["output"]
        .values
        .astype(float)
    )

    # Current best observation
    best_index = np.argmax(y_existing)
    best_point = X_existing[best_index]
    best_output = y_existing[best_index]

    # Scale xi relative to observed output variation
    y_std = np.std(y_existing)

    ei_tuning_results[i] = {}

    print("=" * 70)
    print(f"FUNCTION {i}")
    print("=" * 70)
    print(f"Current best output: {best_output:.6f}")
    print(f"Observed output std: {y_std:.6f}")
    print()

    for xi_prop in xi_proportions:

        xi = xi_prop * y_std

        # Improvement over current best
        improvement = mean_pred - best_output - xi

        # Expected Improvement
        with np.errstate(divide="ignore", invalid="ignore"):
            Z = np.divide(
                improvement,
                std_pred,
                out=np.zeros_like(improvement),
                where=std_pred > 0
            )

            ei_scores = (
                improvement * norm.cdf(Z)
                + std_pred * norm.pdf(Z)
            )

        # EI should be zero where uncertainty is zero
        ei_scores[std_pred <= 0] = 0.0

        # Select highest-EI candidate
        idx = np.argmax(ei_scores)

        point = candidates[idx]
        predicted = mean_pred[idx]
        uncertainty = std_pred[idx]
        score = ei_scores[idx]

        distance_best = np.linalg.norm(
            point - best_point
        )

        distances_existing = np.linalg.norm(
            X_existing - point,
            axis=1
        )

        distance_nearest = distances_existing.min()

        ei_tuning_results[i][xi_prop] = {
            "xi": xi,
            "point": point,
            "predicted": predicted,
            "uncertainty": uncertainty,
            "score": score,
            "distance_best": distance_best,
            "distance_nearest": distance_nearest
        }

        print(
            f"ξ = {xi_prop:.0%} of output std "
            f"(absolute ξ = {xi:.6f})"
        )
        print(np.round(point, 6))
        print(f"Predicted output:               {predicted:.6f}")
        print(f"Uncertainty:                    {uncertainty:.6f}")
        print(f"EI score:                       {score:.6f}")
        print(f"Distance from current best:     {distance_best:.4f}")
        print(f"Distance from nearest existing: {distance_nearest:.4f}")
        print()
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
import numpy as np

nn_models = {}
nn_x_scalers = {}
nn_y_scalers = {}
nn_training_losses = {}

n_ensemble = 5

for i in range(1, 9):

    X = (
        data[i]
        .drop(columns="output")
        .values
        .astype(float)
    )

    y = (
        data[i]["output"]
        .values
        .astype(float)
        .reshape(-1, 1)
    )

    # Standardise inputs and outputs
    x_scaler = StandardScaler()
    y_scaler = StandardScaler()

    X_scaled = x_scaler.fit_transform(X)
    y_scaled = y_scaler.fit_transform(y).ravel()

    models = []
    losses = []

    # Train an ensemble using different random initialisations
    for seed in range(n_ensemble):

        model = MLPRegressor(
            hidden_layer_sizes=(20, 20),
            activation="relu",
            solver="lbfgs",
            alpha=0.001,
            max_iter=5000,
            random_state=seed
        )

        model.fit(X_scaled, y_scaled)

        models.append(model)
        losses.append(model.loss_)

    nn_models[i] = models
    nn_x_scalers[i] = x_scaler
    nn_y_scalers[i] = y_scaler
    nn_training_losses[i] = losses

    print("=" * 60)
    print(f"FUNCTION {i}")
    print("=" * 60)
    print(f"Mean final training loss: {np.mean(losses):.6f}")
    print(f"Minimum final training loss: {np.min(losses):.6f}")
    print(f"Maximum final training loss: {np.max(losses):.6f}")
    print()
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
import numpy as np

nn_loocv_results = {}

for i in range(1, 9):

    X = data[i].drop(columns="output").values.astype(float)
    y = data[i]["output"].values.astype(float)

    nn_predictions = []
    baseline_predictions = []
    actual_values = []

    for j in range(len(X)):

        # Leave one observation out
        mask = np.ones(len(X), dtype=bool)
        mask[j] = False

        X_train = X[mask]
        y_train = y[mask]
        X_test = X[j:j+1]
        y_test = y[j]

        # Scale using training data only
        x_scaler = StandardScaler()
        y_scaler = StandardScaler()

        X_train_scaled = x_scaler.fit_transform(X_train)
        X_test_scaled = x_scaler.transform(X_test)

        y_train_scaled = y_scaler.fit_transform(
            y_train.reshape(-1, 1)
        ).ravel()

        # Train NN
        model = MLPRegressor(
            hidden_layer_sizes=(20, 20),
            activation="relu",
            solver="lbfgs",
            alpha=0.001,
            max_iter=5000,
            random_state=42
        )

        model.fit(X_train_scaled, y_train_scaled)

        # Predict held-out observation
        pred_scaled = model.predict(X_test_scaled)[0]

        pred = y_scaler.inverse_transform(
            [[pred_scaled]]
        )[0, 0]

        nn_predictions.append(pred)
        baseline_predictions.append(np.mean(y_train))
        actual_values.append(y_test)

    # Calculate validation errors
    nn_mae = mean_absolute_error(
        actual_values,
        nn_predictions
    )

    baseline_mae = mean_absolute_error(
        actual_values,
        baseline_predictions
    )

    beats_baseline = nn_mae < baseline_mae

    nn_loocv_results[i] = {
        "nn_mae": nn_mae,
        "baseline_mae": baseline_mae,
        "beats_baseline": beats_baseline
    }

    print("=" * 60)
    print(f"FUNCTION {i}")
    print("=" * 60)
    print(f"NN LOOCV MAE:       {nn_mae:.6f}")
    print(f"Mean baseline MAE:  {baseline_mae:.6f}")
    print(f"NN beats baseline:  {beats_baseline}")
    print()
```

```python
rng_nn = np.random.default_rng(42)

nn_candidates = {}

for i in range(1, 9):

    n_dimensions = data[i].shape[1] - 1

    # Generate candidate points
    candidates = rng_nn.random(
        (100000, n_dimensions)
    )

    # Scale candidates using the scaler fitted in Step 4A
    candidates_scaled = nn_x_scalers[i].transform(
        candidates
    )

    # Predictions from all five NN models
    ensemble_predictions_scaled = np.array([
        model.predict(candidates_scaled)
        for model in nn_models[i]
    ])

    # Convert each model's predictions back to original output scale
    ensemble_predictions = np.array([
        nn_y_scalers[i].inverse_transform(
            preds.reshape(-1, 1)
        ).ravel()
        for preds in ensemble_predictions_scaled
    ])

    # Mean prediction and disagreement across ensemble
    mean_predictions = ensemble_predictions.mean(axis=0)
    disagreement = ensemble_predictions.std(axis=0)

    # Candidate with highest mean predicted output
    idx = np.argmax(mean_predictions)

    selected_point = candidates[idx]
    predicted_output = mean_predictions[idx]
    model_disagreement = disagreement[idx]

    # Existing observations
    X_existing = (
        data[i]
        .drop(columns="output")
        .values
        .astype(float)
    )

    y_existing = (
        data[i]["output"]
        .values
        .astype(float)
    )

    # Current best observation
    best_index = np.argmax(y_existing)
    best_point = X_existing[best_index]

    # Distances
    distance_best = np.linalg.norm(
        selected_point - best_point
    )

    distances_existing = np.linalg.norm(
        X_existing - selected_point,
        axis=1
    )

    distance_nearest = distances_existing.min()

    nn_candidates[i] = {
        "point": selected_point,
        "predicted": predicted_output,
        "disagreement": model_disagreement,
        "distance_best": distance_best,
        "distance_nearest": distance_nearest
    }

    print("=" * 70)
    print(f"FUNCTION {i}")
    print("=" * 70)

    print("NN selected point:")
    print(np.round(selected_point, 6))

    print(
        f"NN predicted output:             "
        f"{predicted_output:.6f}"
    )
    print(
        f"NN model disagreement:          "
        f"{model_disagreement:.6f}"
    )
    print(
        f"Distance from current best:     "
        f"{distance_best:.4f}"
    )
    print(
        f"Distance from nearest existing: "
        f"{distance_nearest:.4f}"
    )

    print()
```

```python
import numpy as np

# Representative acquisition settings
ucb_settings = [0.5, 1.5, 2.5]
ei_settings = [0.00, 0.05, 0.10]

neighbourhood_results = {}

for i in range(1, 9):

    X_existing = (
        data[i]
        .drop(columns="output")
        .values
        .astype(float)
    )

    y_existing = (
        data[i]["output"]
        .values
        .astype(float)
    )

    neighbourhood_results[i] = {}

    print("=" * 75)
    print(f"FUNCTION {i}")
    print("=" * 75)

    # ---------------------------------------------------------
    # UCB candidates
    # ---------------------------------------------------------
    for kappa in ucb_settings:

        point = ucb_tuning_results[i][kappa]["point"]

        distances = np.linalg.norm(
            X_existing - point,
            axis=1
        )

        nearest_idx = np.argsort(distances)[:3]

        key = f"UCB kappa={kappa}"
        neighbourhood_results[i][key] = []

        print(f"\nUCB κ = {kappa}")
        print("Candidate:", np.round(point, 6))
        print("Nearest observations:")

        for idx in nearest_idx:

            neighbourhood_results[i][key].append({
                "distance": distances[idx],
                "output": y_existing[idx]
            })

            print(
                f"  Distance = {distances[idx]:.4f} | "
                f"Output = {y_existing[idx]:.6f}"
            )

    # ---------------------------------------------------------
    # EI candidates
    # ---------------------------------------------------------
    for xi_prop in ei_settings:

        point = ei_tuning_results[i][xi_prop]["point"]

        distances = np.linalg.norm(
            X_existing - point,
            axis=1
        )

        nearest_idx = np.argsort(distances)[:3]

        key = f"EI xi={xi_prop}"
        neighbourhood_results[i][key] = []

        print(f"\nEI ξ = {xi_prop:.0%}")
        print("Candidate:", np.round(point, 6))
        print("Nearest observations:")

        for idx in nearest_idx:

            neighbourhood_results[i][key].append({
                "distance": distances[idx],
                "output": y_existing[idx]
            })

            print(
                f"  Distance = {distances[idx]:.4f} | "
                f"Output = {y_existing[idx]:.6f}"
            )

    # ---------------------------------------------------------
    # NN candidate
    # ---------------------------------------------------------
    point = nn_candidates[i]["point"]

    distances = np.linalg.norm(
        X_existing - point,
        axis=1
    )

    nearest_idx = np.argsort(distances)[:3]

    neighbourhood_results[i]["NN"] = []

    print("\nNeural Network")
    print("Candidate:", np.round(point, 6))
    print("Nearest observations:")

    for idx in nearest_idx:

        neighbourhood_results[i]["NN"].append({
            "distance": distances[idx],
            "output": y_existing[idx]
        })

        print(
            f"  Distance = {distances[idx]:.4f} | "
            f"Output = {y_existing[idx]:.6f}"
        )

    print()
```

```python
# STEP 7A – Generate a novel local-refinement candidate for Function 5

i = 5

X_existing = data[i].drop(columns="output").values.astype(float)
y_existing = data[i]["output"].values.astype(float)

best_idx = np.argmax(y_existing)
best_point = X_existing[best_idx]
best_output = y_existing[best_idx]

rng_local = np.random.default_rng(42)

# Generate points close to the current best
n_local = 100000
local_candidates = best_point + rng_local.normal(
    loc=0.0,
    scale=0.03,
    size=(n_local, len(best_point))
)

# Keep within [0, 1]
local_candidates = np.clip(local_candidates, 0, 1)

# Remove exact/near duplicates
dist_existing = np.min(
    np.linalg.norm(
        local_candidates[:, None, :] - X_existing[None, :, :],
        axis=2
    ),
    axis=1
)

valid = dist_existing > 0.005
local_candidates = local_candidates[valid]
dist_existing = dist_existing[valid]

# GP predictions
gp_mean, gp_std = gp_models[i].predict(
    local_candidates,
    return_std=True
)

# NN ensemble predictions
local_scaled = nn_x_scalers[i].transform(local_candidates)

nn_preds_scaled = np.array([
    model.predict(local_scaled)
    for model in nn_models[i]
])

nn_preds = np.array([
    nn_y_scalers[i].inverse_transform(
        p.reshape(-1, 1)
    ).ravel()
    for p in nn_preds_scaled
])

nn_mean = nn_preds.mean(axis=0)
nn_disagreement = nn_preds.std(axis=0)

# Rank each model's predictions and combine ranks
gp_rank = np.argsort(np.argsort(-gp_mean))
nn_rank = np.argsort(np.argsort(-nn_mean))

combined_rank = gp_rank + nn_rank

idx = np.argmin(combined_rank)

f5_local_candidate = local_candidates[idx]

print("=" * 70)
print("FUNCTION 5 – LOCAL REFINEMENT")
print("=" * 70)

print("Current best:")
print(np.round(best_point, 6))
print(f"Current best output:             {best_output:.6f}")
print()

print("Selected local candidate:")
print(np.round(f5_local_candidate, 6))
print(f"GP predicted output:             {gp_mean[idx]:.6f}")
print(f"GP uncertainty:                  {gp_std[idx]:.6f}")
print(f"NN predicted output:             {nn_mean[idx]:.6f}")
print(f"NN disagreement:                {nn_disagreement[idx]:.6f}")
print(f"Distance from current best:     {np.linalg.norm(f5_local_candidate-best_point):.6f}")
print(f"Distance from nearest existing: {dist_existing[idx]:.6f}")
```
