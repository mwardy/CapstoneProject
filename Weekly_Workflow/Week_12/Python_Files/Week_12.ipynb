```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF
from sklearn.preprocessing import StandardScaler
import warnings

updated_gp_models = {}

print("=" * 70)
print("UPDATED GP MODEL DIAGNOSTICS")
print("=" * 70)

for i in range(1, 9):

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    # Standardise output
    y_scaler = StandardScaler()
    y_scaled = y_scaler.fit_transform(y.reshape(-1, 1)).ravel()

    # Fit GP using same RBF approach as previous rounds
    kernel = RBF(
        length_scale=0.5,
        length_scale_bounds=(1e-5, 1e5)
    )

    gp = GaussianProcessRegressor(
        kernel=kernel,
        alpha=1e-6,
        normalize_y=False,
        n_restarts_optimizer=10,
        random_state=42
    )

    with warnings.catch_warnings(record=True) as caught_warnings:
        warnings.simplefilter("always")
        gp.fit(X, y_scaled)

    updated_gp_models[i] = {
        "model": gp,
        "y_scaler": y_scaler
    }

    warning_text = (
        str(caught_warnings[-1].message)
        if caught_warnings
        else "None"
    )

    print(f"\nFUNCTION {i}")
    print(f"Observations: {len(df)}")
    print(f"Fitted kernel: {gp.kernel_}")
    print(
        f"Log marginal likelihood: "
        f"{gp.log_marginal_likelihood(gp.kernel_.theta):.6f}"
    )
    print(f"Warning: {warning_text}")
```

```python
from sklearn.model_selection import LeaveOneOut
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
from sklearn.pipeline import make_pipeline

print("=" * 70)
print("UPDATED NN LOOCV VALIDATION")
print("=" * 70)

for i in range(1, 9):

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    loo = LeaveOneOut()

    nn_predictions = []
    actual_values = []
    baseline_predictions = []

    for train_index, test_index in loo.split(X):

        X_train = X[train_index]
        X_test = X[test_index]

        y_train = y[train_index]
        y_test = y[test_index]

        # Scale output
        y_scaler = StandardScaler()

        y_train_scaled = y_scaler.fit_transform(
            y_train.reshape(-1, 1)
        ).ravel()

        # Same NN structure used previously
        nn = make_pipeline(
            StandardScaler(),
            MLPRegressor(
                hidden_layer_sizes=(50, 50),
                activation="relu",
                solver="lbfgs",
                max_iter=5000,
                random_state=42
            )
        )

        nn.fit(X_train, y_train_scaled)

        # Predict held-out observation
        pred_scaled = nn.predict(X_test)

        pred = y_scaler.inverse_transform(
            pred_scaled.reshape(-1, 1)
        ).ravel()[0]

        nn_predictions.append(pred)
        actual_values.append(y_test[0])

        # Simple mean baseline using training data only
        baseline_predictions.append(np.mean(y_train))

    nn_mae = mean_absolute_error(
        actual_values,
        nn_predictions
    )

    baseline_mae = mean_absolute_error(
        actual_values,
        baseline_predictions
    )

    beats_baseline = nn_mae < baseline_mae

    print(f"\nFUNCTION {i}")
    print(f"Observations:       {len(y)}")
    print(f"NN LOOCV MAE:       {nn_mae:.10f}")
    print(f"Baseline MAE:        {baseline_mae:.10f}")
    print(f"Beats baseline:      {beats_baseline}")
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import numpy as np

pca_results = {}

print("=" * 80)
print("PCA EXPLAINED VARIANCE ANALYSIS")
print("=" * 80)

for i in range(1, 9):

    df = data[i]

    # Inputs only - output is deliberately excluded from PCA
    X = df.iloc[:, :-1].values

    # Scale inputs before PCA
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Fit PCA using all available components
    pca = PCA()
    X_pca = pca.fit_transform(X_scaled)

    explained = pca.explained_variance_ratio_
    cumulative = np.cumsum(explained)

    # Components required to reach each threshold
    n_80 = np.searchsorted(cumulative, 0.80) + 1
    n_90 = np.searchsorted(cumulative, 0.90) + 1
    n_95 = np.searchsorted(cumulative, 0.95) + 1

    # Store results for later steps
    pca_results[i] = {
        "pca": pca,
        "scaler": scaler,
        "scores": X_pca,
        "explained_variance": explained,
        "cumulative_variance": cumulative
    }

    print(f"\nFUNCTION {i}")
    print(f"Observations: {len(X)}")
    print(f"Original dimensions: {X.shape[1]}")

    print("\nExplained variance:")
    for j, variance in enumerate(explained):
        print(
            f"  PC{j+1}: "
            f"{variance:.4f} "
            f"({variance * 100:.2f}%)"
        )

    print("\nCumulative explained variance:")
    for j, variance in enumerate(cumulative):
        print(
            f"  PC1-PC{j+1}: "
            f"{variance:.4f} "
            f"({variance * 100:.2f}%)"
        )

    print("\nComponents required:")
    print(f"  80% variance: {n_80}")
    print(f"  90% variance: {n_90}")
    print(f"  95% variance: {n_95}")
```

```python
print("=" * 85)
print("PCA LOADINGS AND OUTPUT RELATIONSHIPS")
print("=" * 85)

pca_performance = {}

for i in range(1, 9):

    df = data[i]
    y = df.iloc[:, -1].values

    pca = pca_results[i]["pca"]
    scores = pca_results[i]["scores"]
    explained = pca_results[i]["explained_variance"]

    n_components = scores.shape[1]
    n_dimensions = df.shape[1] - 1

    # PCA loadings
    loadings = pca.components_

    # Correlation between each PC score and output
    correlations = []

    for j in range(n_components):

        corr = np.corrcoef(scores[:, j], y)[0, 1]
        correlations.append(corr)

    correlations = np.array(correlations)

    # Component with strongest absolute relationship with output
    strongest_pc = np.argmax(np.abs(correlations))

    pca_performance[i] = {
        "loadings": loadings,
        "correlations": correlations,
        "strongest_pc": strongest_pc
    }

    print(f"\n{'=' * 85}")
    print(f"FUNCTION {i}")
    print(f"{'=' * 85}")

    print("\nPrincipal component loadings:")

    for j in range(n_components):

        print(
            f"\nPC{j+1} "
            f"({explained[j] * 100:.2f}% variance)"
        )

        for d in range(n_dimensions):
            print(
                f"  Dimension {d+1}: "
                f"{loadings[j, d]: .4f}"
            )

    print("\nRelationship with output:")

    for j, corr in enumerate(correlations):
        print(
            f"  PC{j+1} correlation with output: "
            f"{corr:.4f}"
        )

    print(
        f"\nStrongest output relationship: "
        f"PC{strongest_pc + 1} "
        f"(correlation = "
        f"{correlations[strongest_pc]:.4f})"
    )
```

```python
print("=" * 90)
print("HIGH-PERFORMING OBSERVATIONS IN PCA SPACE")
print("=" * 90)

pca_top_regions = {}

for i in range(1, 9):

    df = data[i]
    y = df.iloc[:, -1].values

    scores = pca_results[i]["scores"]
    correlations = pca_performance[i]["correlations"]

    # Component with strongest relationship with output
    strongest_pc = np.argmax(np.abs(correlations))

    # Number of observations in top 20%
    n_top = max(3, int(np.ceil(len(y) * 0.20)))

    # Indices of highest-performing observations
    top_indices = np.argsort(y)[-n_top:]

    # Current best observation
    best_index = np.argmax(y)

    # Scores for strongest PC
    pc_scores = scores[:, strongest_pc]

    overall_mean = np.mean(pc_scores)
    overall_std = np.std(pc_scores)

    top_mean = np.mean(pc_scores[top_indices])
    top_min = np.min(pc_scores[top_indices])
    top_max = np.max(pc_scores[top_indices])

    best_score = pc_scores[best_index]

    # Standardised position of current best
    if overall_std > 0:
        best_z = (best_score - overall_mean) / overall_std
        top_mean_z = (top_mean - overall_mean) / overall_std
    else:
        best_z = 0
        top_mean_z = 0

    pca_top_regions[i] = {
        "strongest_pc": strongest_pc,
        "top_indices": top_indices,
        "top_mean": top_mean,
        "best_score": best_score,
        "best_z": best_z,
        "top_mean_z": top_mean_z
    }

    print(f"\n{'=' * 90}")
    print(f"FUNCTION {i}")
    print(f"{'=' * 90}")

    print(f"Strongest component:        PC{strongest_pc + 1}")
    print(f"Output correlation:         {correlations[strongest_pc]:.4f}")
    print(f"Top observations analysed:  {n_top}")

    print(f"\nOverall PC mean:             {overall_mean:.4f}")
    print(f"Top 20% PC mean:             {top_mean:.4f}")
    print(f"Top 20% PC range:            {top_min:.4f} to {top_max:.4f}")

    print(f"\nCurrent best index:          {best_index}")
    print(f"Current best PC score:       {best_score:.4f}")
    print(f"Current best z-position:     {best_z:.4f}")
    print(f"Top 20% mean z-position:     {top_mean_z:.4f}")

    # Simple interpretation
    if correlations[strongest_pc] > 0:
        expected_direction = "HIGHER"
    else:
        expected_direction = "LOWER"

    print(
        f"\nHigher outputs associated with: "
        f"{expected_direction} PC{strongest_pc + 1} scores"
    )
```

```python
# ================================================================
# STEP 8 - PCA-GUIDED CANDIDATE GENERATION
# ================================================================

from scipy.spatial.distance import cdist
import numpy as np

print("=" * 90)
print("PCA-GUIDED CANDIDATE GENERATION")
print("=" * 90)

# Functions where PCA showed enough evidence to be useful
pca_functions = [2, 4, 5, 6, 7, 8]

# Strength of PCA evidence
pca_strength = {
    2: "Moderate",
    4: "Moderate",
    5: "Strong",
    6: "Strong",
    7: "Strong",
    8: "Strong"
}

# Small movements in standardised PCA space
step_sizes = [0.10, 0.20, 0.30]

pca_candidates = {}

for i in pca_functions:

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    scaler = pca_results[i]["scaler"]
    pca = pca_results[i]["pca"]

    correlations = pca_performance[i]["correlations"]

    # Current best observation
    best_index = np.argmax(y)
    best_point = X[best_index]

    # Strongest performance-related PC
    pc_index = np.argmax(np.abs(correlations))
    correlation = correlations[pc_index]

    # Direction associated with better outputs
    direction = 1 if correlation > 0 else -1

    # Convert current best into PCA space
    best_scaled = scaler.transform(
        best_point.reshape(1, -1)
    )

    best_pca = pca.transform(best_scaled)[0]

    print(f"\n{'=' * 90}")
    print(f"FUNCTION {i}")
    print(f"{'=' * 90}")

    print(f"PCA evidence:             {pca_strength[i]}")
    print(f"Current best output:      {y[best_index]:.10f}")
    print(
        f"Performance component:    PC{pc_index + 1}"
    )
    print(
        f"Output correlation:       {correlation:.4f}"
    )
    print(
        f"Favourable direction:     "
        f"{'HIGHER' if direction > 0 else 'LOWER'} "
        f"PC{pc_index + 1}"
    )

    function_candidates = []

    for step in step_sizes:

        # Copy current best PCA coordinates
        new_pca = best_pca.copy()

        # Move only along selected component
        new_pca[pc_index] += direction * step

        # Transform back to original input space
        candidate_scaled = pca.inverse_transform(
            new_pca.reshape(1, -1)
        )

        candidate = scaler.inverse_transform(
            candidate_scaled
        )[0]

        # Keep candidate inside valid bounds
        candidate = np.clip(candidate, 0, 1)

        # Distance from existing observations
        nearest_distance = np.min(
            cdist(
                candidate.reshape(1, -1),
                X
            )
        )

        function_candidates.append({
            "step": step,
            "candidate": candidate,
            "nearest_distance": nearest_distance
        })

        print(f"\nPCA step: {step:.2f}")
        print(
            "Candidate:        "
            + np.array2string(
                candidate,
                precision=6,
                separator=", "
            )
        )
        print(
            f"Nearest distance: {nearest_distance:.6f}"
        )

    pca_candidates[i] = function_candidates
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
import numpy as np

print("=" * 90)
print("PCA CANDIDATE MODEL EVALUATION")
print("=" * 90)

pca_candidate_evaluation = {}

for i in [2, 4, 6, 7, 8]:

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    # ------------------------------------------------------------
    # Updated GP from Step 3
    # ------------------------------------------------------------
    gp = updated_gp_models[i]["model"]
    gp_y_scaler = updated_gp_models[i]["y_scaler"]

    # ------------------------------------------------------------
    # Fit NN to full updated dataset
    # ------------------------------------------------------------
    nn_y_scaler = StandardScaler()

    y_scaled = nn_y_scaler.fit_transform(
        y.reshape(-1, 1)
    ).ravel()

    nn = make_pipeline(
        StandardScaler(),
        MLPRegressor(
            hidden_layer_sizes=(50, 50),
            activation="relu",
            solver="lbfgs",
            max_iter=5000,
            random_state=42
        )
    )

    nn.fit(X, y_scaled)

    print(f"\n{'=' * 90}")
    print(f"FUNCTION {i}")
    print(f"{'=' * 90}")
    print(f"Current best output: {np.max(y):.10f}")

    evaluations = []

    for result in pca_candidates[i]:

        step = result["step"]
        candidate = result["candidate"]
        distance = result["nearest_distance"]

        # GP prediction
        gp_mean_scaled, gp_std_scaled = gp.predict(
            candidate.reshape(1, -1),
            return_std=True
        )

        gp_prediction = gp_y_scaler.inverse_transform(
            gp_mean_scaled.reshape(-1, 1)
        ).ravel()[0]

        # Convert GP uncertainty back to original output scale
        gp_uncertainty = (
            gp_std_scaled[0] *
            gp_y_scaler.scale_[0]
        )

        # NN prediction
        nn_pred_scaled = nn.predict(
            candidate.reshape(1, -1)
        )

        nn_prediction = nn_y_scaler.inverse_transform(
            nn_pred_scaled.reshape(-1, 1)
        ).ravel()[0]

        evaluations.append({
            "step": step,
            "candidate": candidate,
            "gp_prediction": gp_prediction,
            "gp_uncertainty": gp_uncertainty,
            "nn_prediction": nn_prediction,
            "distance": distance
        })

        print(f"\nPCA step:          {step:.2f}")

        print(
            "Candidate:         "
            + np.array2string(
                candidate,
                precision=6,
                separator=", "
            )
        )

        print(
            f"GP predicted:      "
            f"{gp_prediction:.10f}"
        )

        print(
            f"GP uncertainty:    "
            f"{gp_uncertainty:.10f}"
        )

        print(
            f"NN predicted:      "
            f"{nn_prediction:.10f}"
        )

        print(
            f"Nearest distance:  "
            f"{distance:.6f}"
        )

    pca_candidate_evaluation[i] = evaluations
```

```python
from scipy.spatial.distance import cdist
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
import numpy as np

np.random.seed(42)

print("=" * 90)
print("FUNCTION-SPECIFIC ROUND 13 CANDIDATE REFINEMENT")
print("=" * 90)

final_candidates = {}

# ------------------------------------------------------------
# Helper: NN prediction
# ------------------------------------------------------------

def fit_nn_predictor(X, y):

    y_scaler = StandardScaler()
    y_scaled = y_scaler.fit_transform(
        y.reshape(-1, 1)
    ).ravel()

    nn = make_pipeline(
        StandardScaler(),
        MLPRegressor(
            hidden_layer_sizes=(50, 50),
            activation="relu",
            solver="lbfgs",
            max_iter=5000,
            random_state=42
        )
    )

    nn.fit(X, y_scaled)

    return nn, y_scaler


# ------------------------------------------------------------
# FUNCTION 1
# Model-independent exploration
# ------------------------------------------------------------

i = 1

df = data[i]
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

candidates = np.random.rand(100000, X.shape[1])

distances = cdist(candidates, X).min(axis=1)

idx = np.argmax(distances)

candidate = candidates[idx]

final_candidates[i] = candidate

print("\nFUNCTION 1 - MODEL-INDEPENDENT EXPLORATION")
print(f"Current best:      {np.max(y):.10f}")
print(f"Candidate:         {np.round(candidate, 6)}")
print(f"Nearest distance:  {distances[idx]:.6f}")


# ------------------------------------------------------------
# FUNCTIONS 2-8
# ------------------------------------------------------------

for i in range(2, 9):

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    best_idx = np.argmax(y)
    best_point = X[best_idx]

    # Fit NN
    nn, nn_y_scaler = fit_nn_predictor(X, y)

    # Updated GP
    gp = updated_gp_models[i]["model"]
    gp_y_scaler = updated_gp_models[i]["y_scaler"]

    # --------------------------------------------------------
    # Generate local candidates
    # --------------------------------------------------------

    if i == 5:
        # F5: targeted upper-boundary search
        candidates = np.random.uniform(
            low=0.94,
            high=1.0,
            size=(100000, X.shape[1])
        )

    else:
        # Local search around current best
        radius = {
            2: 0.10,
            3: 0.07,
            4: 0.08,
            6: 0.08,
            7: 0.08,
            8: 0.08
        }[i]

        candidates = (
            best_point
            + np.random.uniform(
                -radius,
                radius,
                size=(100000, X.shape[1])
            )
        )

        candidates = np.clip(candidates, 0, 1)

    # --------------------------------------------------------
    # Distance from existing observations
    # --------------------------------------------------------

    distances = cdist(candidates, X).min(axis=1)

    # Minimum novelty
    min_distance = {
        2: 0.02,
        3: 0.025,
        4: 0.03,
        5: 0.02,
        6: 0.03,
        7: 0.04,
        8: 0.05
    }[i]

    valid = distances >= min_distance

    candidates_valid = candidates[valid]
    distances_valid = distances[valid]

    # --------------------------------------------------------
    # GP predictions
    # --------------------------------------------------------

    gp_scaled, gp_std_scaled = gp.predict(
        candidates_valid,
        return_std=True
    )

    gp_pred = gp_y_scaler.inverse_transform(
        gp_scaled.reshape(-1, 1)
    ).ravel()

    gp_std = (
        gp_std_scaled
        * gp_y_scaler.scale_[0]
    )

    # --------------------------------------------------------
    # NN predictions
    # --------------------------------------------------------

    nn_scaled = nn.predict(candidates_valid)

    nn_pred = nn_y_scaler.inverse_transform(
        nn_scaled.reshape(-1, 1)
    ).ravel()

    # --------------------------------------------------------
    # Function-specific scoring
    # --------------------------------------------------------

    if i == 2:
        # Broken GP -> NN only
        score = nn_pred

        method = "NN-LED LOCAL REFINEMENT"

    elif i == 3:
        # Mixed evidence -> balanced GP/NN
        score = (
            0.50 * gp_pred
            + 0.50 * nn_pred
        )

        method = "BALANCED GP-NN LOCAL REFINEMENT"

    elif i == 4:
        # Latest realised result favours NN
        score = (
            0.30 * gp_pred
            + 0.70 * nn_pred
        )

        method = "NN-WEIGHTED LOCAL REFINEMENT"

    elif i == 5:
        # Strong boundary evidence
        score = (
            0.60 * gp_pred
            + 0.40 * nn_pred
        )

        method = "TARGETED BOUNDARY REFINEMENT"

    elif i == 6:
        # GP recently stronger; PCA supports region
        score = (
            0.65 * gp_pred
            + 0.35 * nn_pred
        )

        method = "GP-WEIGHTED PCA-SUPPORTED REFINEMENT"

    elif i == 7:
        # Strong NN + PCA evidence
        score = (
            0.40 * gp_pred
            + 0.60 * nn_pred
        )

        method = "PCA-SUPPORTED GP-NN REFINEMENT"

    elif i == 8:
        # Strong NN/PCA, but retain GP influence
        score = (
            0.30 * gp_pred
            + 0.70 * nn_pred
        )

        method = "PCA-NN WEIGHTED REFINEMENT"

    # --------------------------------------------------------
    # Select best candidate
    # --------------------------------------------------------

    idx = np.argmax(score)

    candidate = candidates_valid[idx]

    final_candidates[i] = candidate

    print(f"\nFUNCTION {i} - {method}")

    print(
        f"Current best:      "
        f"{np.max(y):.10f}"
    )

    print(
        "Candidate:         "
        + np.array2string(
            candidate,
            precision=6,
            separator=", "
        )
    )

    print(
        f"GP predicted:      "
        f"{gp_pred[idx]:.10f}"
    )

    print(
        f"GP uncertainty:    "
        f"{gp_std[idx]:.10f}"
    )

    print(
        f"NN predicted:      "
        f"{nn_pred[idx]:.10f}"
    )

    print(
        f"Nearest distance:  "
        f"{distances_valid[idx]:.6f}"
    )
```

```python
print("=" * 90)
print("FINAL TARGETED CANDIDATE COMPARISON")
print("=" * 90)

targeted_candidates = {}

# ============================================================
# FUNCTION 5 - CONTROLLED BOUNDARY ALTERNATIVES
# ============================================================

i = 5

df = data[i]
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

gp = updated_gp_models[i]["model"]
gp_y_scaler = updated_gp_models[i]["y_scaler"]

nn, nn_y_scaler = fit_nn_predictor(X, y)

# Existing Step 10 candidate plus slightly different
# distances from the known [1,1,1,1] optimum
f5_candidates = np.array([
    final_candidates[5],
    [0.995, 0.995, 0.980, 0.990],
    [0.990, 0.995, 0.980, 0.985],
    [0.985, 0.990, 0.975, 0.985]
])

targeted_candidates[5] = f5_candidates


# ============================================================
# FUNCTIONS 6-8 - PCA/MODEL INTERPOLATION
# ============================================================

# Use PCA 0.20 candidate as conservative starting point
for i in [6, 7, 8]:

    pca_point = pca_candidates[i][1]["candidate"]
    model_point = final_candidates[i]

    # PCA point, two intermediate points, model point
    weights = [0.0, 0.33, 0.67, 1.0]

    candidates = []

    for w in weights:

        candidate = (
            (1 - w) * pca_point
            + w * model_point
        )

        candidate = np.clip(candidate, 0, 1)

        candidates.append(candidate)

    targeted_candidates[i] = np.array(candidates)


# ============================================================
# EVALUATE F5-F8
# ============================================================

for i in [5, 6, 7, 8]:

    df = data[i]

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    gp = updated_gp_models[i]["model"]
    gp_y_scaler = updated_gp_models[i]["y_scaler"]

    nn, nn_y_scaler = fit_nn_predictor(X, y)

    candidates = targeted_candidates[i]

    distances = cdist(candidates, X).min(axis=1)

    # GP
    gp_scaled, gp_std_scaled = gp.predict(
        candidates,
        return_std=True
    )

    gp_pred = gp_y_scaler.inverse_transform(
        gp_scaled.reshape(-1, 1)
    ).ravel()

    gp_std = (
        gp_std_scaled
        * gp_y_scaler.scale_[0]
    )

    # NN
    nn_scaled = nn.predict(candidates)

    nn_pred = nn_y_scaler.inverse_transform(
        nn_scaled.reshape(-1, 1)
    ).ravel()

    print(f"\n{'=' * 90}")
    print(f"FUNCTION {i}")
    print(f"{'=' * 90}")

    print(f"Current best: {np.max(y):.10f}")

    for j, candidate in enumerate(candidates):

        if i == 5:
            label = f"Boundary option {j + 1}"
        else:
            weights = [0.0, 0.33, 0.67, 1.0]
            label = f"PCA → model weight {weights[j]:.2f}"

        print(f"\n{label}")

        print(
            "Candidate:        "
            + np.array2string(
                candidate,
                precision=6,
                separator=", "
            )
        )

        print(
            f"GP predicted:     "
            f"{gp_pred[j]:.10f}"
        )

        print(
            f"GP uncertainty:   "
            f"{gp_std[j]:.10f}"
        )

        print(
            f"NN predicted:     "
            f"{nn_pred[j]:.10f}"
        )

        print(
            f"Nearest distance: "
            f"{distances[j]:.6f}"
        )
```
