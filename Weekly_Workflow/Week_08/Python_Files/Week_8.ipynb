```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF

gp_models = {}

for i in range(1, 9):

    # Separate inputs and outputs
    X = data[i].drop(columns="output").values
    y = data[i]["output"].values

    # Fit GP using the established RBF configuration
    gp = GaussianProcessRegressor(
        kernel=RBF(length_scale=1.0, length_scale_bounds=(1e-5, 1e5)),
        normalize_y=True,
        n_restarts_optimizer=10,
        random_state=42
    )

    gp.fit(X, y)
    gp_models[i] = gp

    print("\n" + "=" * 60)
    print(f"FUNCTION {i}")
    print("=" * 60)
    print(f"Observations: {len(X)}")
    print(f"Fitted kernel: {gp.kernel_}")
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler

nn_models = {}
nn_x_scalers = {}
nn_y_scalers = {}
nn_training_losses = {}

n_ensemble = 5

for i in range(1, 9):

    X = (
        data[i]
        .drop(columns="output")
        .values
        .astype(float)
    )

    y = (
        data[i]["output"]
        .values
        .astype(float)
        .reshape(-1, 1)
    )

    # Standardise inputs and outputs
    x_scaler = StandardScaler()
    y_scaler = StandardScaler()

    X_scaled = x_scaler.fit_transform(X)
    y_scaled = y_scaler.fit_transform(y).ravel()

    models = []
    losses = []

    # Train an ensemble using different random initialisations
    for seed in range(n_ensemble):

        model = MLPRegressor(
            hidden_layer_sizes=(20, 20),
            activation="relu",
            solver="lbfgs",
            alpha=0.001,
            max_iter=5000,
            random_state=seed
        )

        model.fit(X_scaled, y_scaled)

        models.append(model)
        losses.append(model.loss_)

    nn_models[i] = models
    nn_x_scalers[i] = x_scaler
    nn_y_scalers[i] = y_scaler
    nn_training_losses[i] = losses

    print("=" * 60)
    print(f"FUNCTION {i}")
    print("=" * 60)
    print(f"Mean final training loss: {np.mean(losses):.6f}")
    print(f"Minimum final training loss: {np.min(losses):.6f}")
    print(f"Maximum final training loss: {np.max(losses):.6f}")
    print()
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
import numpy as np

nn_loocv_results = {}

for i in range(1, 9):

    X = data[i].drop(columns="output").values.astype(float)
    y = data[i]["output"].values.astype(float)

    nn_predictions = []
    baseline_predictions = []
    actual_values = []

    for j in range(len(X)):

        # Leave one observation out
        mask = np.ones(len(X), dtype=bool)
        mask[j] = False

        X_train = X[mask]
        y_train = y[mask]
        X_test = X[j:j+1]
        y_test = y[j]

        # Scale using training data only
        x_scaler = StandardScaler()
        y_scaler = StandardScaler()

        X_train_scaled = x_scaler.fit_transform(X_train)
        X_test_scaled = x_scaler.transform(X_test)

        y_train_scaled = y_scaler.fit_transform(
            y_train.reshape(-1, 1)
        ).ravel()

        # Train NN using the established configuration
        model = MLPRegressor(
            hidden_layer_sizes=(20, 20),
            activation="relu",
            solver="lbfgs",
            alpha=0.001,
            max_iter=5000,
            random_state=42
        )

        model.fit(X_train_scaled, y_train_scaled)

        # Predict the held-out observation
        pred_scaled = model.predict(X_test_scaled)[0]

        pred = y_scaler.inverse_transform(
            [[pred_scaled]]
        )[0, 0]

        nn_predictions.append(pred)
        baseline_predictions.append(np.mean(y_train))
        actual_values.append(y_test)

    # Calculate validation errors
    nn_mae = mean_absolute_error(
        actual_values,
        nn_predictions
    )

    baseline_mae = mean_absolute_error(
        actual_values,
        baseline_predictions
    )

    beats_baseline = nn_mae < baseline_mae

    nn_loocv_results[i] = {
        "nn_mae": nn_mae,
        "baseline_mae": baseline_mae,
        "beats_baseline": beats_baseline
    }

    print("=" * 60)
    print(f"FUNCTION {i}")
    print("=" * 60)
    print(f"NN LOOCV MAE:       {nn_mae:.6f}")
    print(f"Mean baseline MAE:  {baseline_mae:.6f}")
    print(f"NN beats baseline:  {beats_baseline}")
    print()
```

```python
from scipy.stats import norm
from scipy.spatial.distance import cdist
import numpy as np

# Representative settings based on Week 7 analysis
kappa_values = [0.5, 1.5]
xi_percentages = [0.00, 0.05]

gp_candidate_results = {}

for i in range(1, 9):

    X = data[i].drop(columns="output").values.astype(float)
    y = data[i]["output"].values.astype(float)

    candidates = candidate_sets[i]

    # GP predictions
    mean, std = gp_models[i].predict(
        candidates,
        return_std=True
    )

    current_best = np.max(y)
    y_std = np.std(y)

    results = {
        "UCB": {},
        "EI": {}
    }

    print("\n" + "=" * 70)
    print(f"FUNCTION {i}")
    print("=" * 70)
    print(f"Current best output: {current_best:.6f}")

    # ---------------------------------------------------------
    # UCB
    # ---------------------------------------------------------
    print("\nUCB CANDIDATES")

    for kappa in kappa_values:

        ucb = mean + kappa * std
        idx = np.argmax(ucb)

        point = candidates[idx]
        nearest_distance = np.min(
            cdist(point.reshape(1, -1), X)
        )

        results["UCB"][kappa] = {
            "point": point,
            "predicted": mean[idx],
            "uncertainty": std[idx],
            "score": ucb[idx],
            "nearest_distance": nearest_distance
        }

        print(f"\nκ = {kappa}")
        print(np.round(point, 6))
        print(f"Predicted output:              {mean[idx]:.6f}")
        print(f"Uncertainty:                   {std[idx]:.6f}")
        print(f"UCB score:                     {ucb[idx]:.6f}")
        print(f"Distance from nearest existing:{nearest_distance:.6f}")

    # ---------------------------------------------------------
    # EI
    # ---------------------------------------------------------
    print("\nEI CANDIDATES")

    for xi_pct in xi_percentages:

        xi = xi_pct * y_std
        improvement = mean - current_best - xi

        with np.errstate(divide="ignore", invalid="ignore"):
            Z = np.divide(
                improvement,
                std,
                out=np.zeros_like(improvement),
                where=std > 0
            )

        ei = improvement * norm.cdf(Z) + std * norm.pdf(Z)
        ei[std == 0] = 0

        idx = np.argmax(ei)

        point = candidates[idx]
        nearest_distance = np.min(
            cdist(point.reshape(1, -1), X)
        )

        results["EI"][xi_pct] = {
            "point": point,
            "predicted": mean[idx],
            "uncertainty": std[idx],
            "score": ei[idx],
            "nearest_distance": nearest_distance
        }

        print(f"\nξ = {int(xi_pct * 100)}%")
        print(np.round(point, 6))
        print(f"Predicted output:              {mean[idx]:.6f}")
        print(f"Uncertainty:                   {std[idx]:.6f}")
        print(f"EI score:                      {ei[idx]:.6f}")
        print(f"Distance from nearest existing:{nearest_distance:.6f}")

    gp_candidate_results[i] = results
```

```python
from scipy.spatial.distance import cdist

nn_candidate_results = {}

for i in range(1, 9):

    X = data[i].drop(columns="output").values.astype(float)
    y = data[i]["output"].values.astype(float)

    candidates = candidate_sets[i]

    # Scale candidate inputs using the scaler fitted in Step 2
    candidates_scaled = nn_x_scalers[i].transform(candidates)

    # Generate predictions from each of the five NN models
    model_predictions = []

    for model in nn_models[i]:

        pred_scaled = model.predict(candidates_scaled)

        # Convert predictions back to original output scale
        pred = nn_y_scalers[i].inverse_transform(
            pred_scaled.reshape(-1, 1)
        ).ravel()

        model_predictions.append(pred)

    model_predictions = np.array(model_predictions)

    # Mean prediction and disagreement across ensemble
    mean_prediction = np.mean(model_predictions, axis=0)
    model_disagreement = np.std(model_predictions, axis=0)

    # Select candidate with highest mean NN prediction
    idx = np.argmax(mean_prediction)

    selected_point = candidates[idx]
    selected_prediction = mean_prediction[idx]
    selected_disagreement = model_disagreement[idx]

    # Distance from current best observation
    best_idx = np.argmax(y)
    current_best_point = X[best_idx]

    distance_from_best = np.linalg.norm(
        selected_point - current_best_point
    )

    # Distance from nearest existing observation
    nearest_distance = np.min(
        cdist(selected_point.reshape(1, -1), X)
    )

    nn_candidate_results[i] = {
        "point": selected_point,
        "predicted": selected_prediction,
        "disagreement": selected_disagreement,
        "distance_from_best": distance_from_best,
        "nearest_distance": nearest_distance
    }

    print("\n" + "=" * 70)
    print(f"FUNCTION {i}")
    print("=" * 70)

    print("NN selected point:")
    print(np.round(selected_point, 6))

    print(f"NN predicted output:             {selected_prediction:.6f}")
    print(f"NN model disagreement:          {selected_disagreement:.6f}")
    print(f"Distance from current best:     {distance_from_best:.4f}")
    print(f"Distance from nearest existing: {nearest_distance:.4f}")
```

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, Matern, WhiteKernel
from sklearn.metrics import mean_absolute_error
import numpy as np

# Function 1 data
X = data[1].drop(columns="output").values.astype(float)
y = data[1]["output"].values.astype(float)

kernels = {
    "RBF": RBF(
        length_scale=0.1,
        length_scale_bounds=(1e-5, 1e2)
    ),

    "Matern 3/2": Matern(
        length_scale=0.1,
        length_scale_bounds=(1e-5, 1e2),
        nu=1.5
    ),

    "Matern 5/2": Matern(
        length_scale=0.1,
        length_scale_bounds=(1e-5, 1e2),
        nu=2.5
    ),

    "RBF + noise": (
        RBF(
            length_scale=0.1,
            length_scale_bounds=(1e-5, 1e2)
        )
        + WhiteKernel(
            noise_level=1e-5,
            noise_level_bounds=(1e-8, 1e0)
        )
    )
}

print("=" * 70)
print("FUNCTION 1 – ALTERNATIVE GP KERNEL LOOCV")
print("=" * 70)

for name, kernel in kernels.items():

    predictions = []
    actual = []
    baseline_predictions = []

    for j in range(len(X)):

        mask = np.ones(len(X), dtype=bool)
        mask[j] = False

        X_train = X[mask]
        y_train = y[mask]

        X_test = X[j:j+1]
        y_test = y[j]

        gp = GaussianProcessRegressor(
            kernel=kernel,
            normalize_y=True,
            n_restarts_optimizer=10,
            random_state=42
        )

        gp.fit(X_train, y_train)

        pred = gp.predict(X_test)[0]

        predictions.append(pred)
        actual.append(y_test)
        baseline_predictions.append(np.mean(y_train))

    gp_mae = mean_absolute_error(actual, predictions)
    baseline_mae = mean_absolute_error(
        actual,
        baseline_predictions
    )

    # Fit full dataset so we can inspect final kernel
    final_gp = GaussianProcessRegressor(
        kernel=kernel,
        normalize_y=True,
        n_restarts_optimizer=10,
        random_state=42
    )

    final_gp.fit(X, y)

    print(f"\n{name}")
    print(f"LOOCV MAE:          {gp_mae:.10f}")
    print(f"Baseline MAE:       {baseline_mae:.10f}")
    print(f"Beats baseline:     {gp_mae < baseline_mae}")
    print(f"Final fitted kernel: {final_gp.kernel_}")
```

```python
from scipy.spatial.distance import cdist
import numpy as np

rng = np.random.default_rng(42)

local_results = {}

for i in [4, 5, 7]:

    X = data[i].drop(columns="output").values.astype(float)
    y = data[i]["output"].values.astype(float)

    # Current best observation
    best_idx = np.argmax(y)
    best_point = X[best_idx]
    best_output = y[best_idx]

    # Generate local candidates around current best
    n_candidates = 100000
    local_scale = 0.08

    candidates = best_point + rng.normal(
        0,
        local_scale,
        size=(n_candidates, X.shape[1])
    )

    # Keep candidates inside permitted bounds
    candidates = np.clip(candidates, 0, 1)

    # Remove candidates effectively duplicating existing observations
    distances = cdist(candidates, X)
    nearest_distance = np.min(distances, axis=1)

    valid = nearest_distance > 1e-6

    candidates = candidates[valid]
    nearest_distance = nearest_distance[valid]

    # GP predictions
    gp_mean, gp_std = gp_models[i].predict(
        candidates,
        return_std=True
    )

    # NN predictions from ensemble
    candidates_scaled = nn_x_scalers[i].transform(candidates)

    nn_predictions = []

    for model in nn_models[i]:

        pred_scaled = model.predict(candidates_scaled)

        pred = nn_y_scalers[i].inverse_transform(
            pred_scaled.reshape(-1, 1)
        ).ravel()

        nn_predictions.append(pred)

    nn_predictions = np.array(nn_predictions)

    nn_mean = np.mean(nn_predictions, axis=0)
    nn_disagreement = np.std(nn_predictions, axis=0)

    # Combined standardised GP + NN prediction
    gp_z = (
        (gp_mean - np.mean(gp_mean)) /
        (np.std(gp_mean) + 1e-12)
    )

    nn_z = (
        (nn_mean - np.mean(nn_mean)) /
        (np.std(nn_mean) + 1e-12)
    )

    combined_score = (gp_z + nn_z) / 2

    best_candidate_idx = np.argmax(combined_score)

    selected = candidates[best_candidate_idx]

    local_results[i] = selected

    print("=" * 70)
    print(f"FUNCTION {i} – LOCAL REFINEMENT")
    print("=" * 70)

    print("Current best:")
    print(np.round(best_point, 6))
    print(f"Current best output:             {best_output:.6f}")

    print("\nSelected local candidate:")
    print(np.round(selected, 6))

    print(
        f"GP predicted output:             "
        f"{gp_mean[best_candidate_idx]:.6f}"
    )

    print(
        f"GP uncertainty:                  "
        f"{gp_std[best_candidate_idx]:.6f}"
    )

    print(
        f"NN predicted output:             "
        f"{nn_mean[best_candidate_idx]:.6f}"
    )

    print(
        f"NN disagreement:                 "
        f"{nn_disagreement[best_candidate_idx]:.6f}"
    )

    print(
        f"Distance from current best:      "
        f"{np.linalg.norm(selected - best_point):.6f}"
    )

    print(
        f"Distance from nearest existing:  "
        f"{nearest_distance[best_candidate_idx]:.6f}"
    )

    print()
```
